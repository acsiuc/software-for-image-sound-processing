```python
import datetime as dt

my_birthday = dt.date(2004, 4, 10)
my_age = 22
birthdays_on_wednesday = 0

for x in range(my_age):
    if my_birthday.weekday() == 2:
        birthdays_on_wednesday +=1
    my_birthday = my_birthday.replace(year = my_birthday.year + 1)

    
    
print(f"My birthday was on wednesday {birthdays_on_wednesday} times")



```
