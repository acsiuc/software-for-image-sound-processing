```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
import unicodedata

fs = 1000
coefficients = np.ones(5)/5
w,h = signal.freqz(coefficients)
magnitude = np.abs(h)
cutoff_index = np.where(magnitude <= 0.707)[0][0]
cutoff_omega = w[cutoff_index]
cutoff_freq = (cutoff_omega * fs)/(2*np.pi)
plt.plot(w, 20 * np.log10(magnitude), 'b')
plt.ylabel('Amplitude[dB]', color ='b')
plt.xlabel(unicodedata.lookup("GREEK CAPITAL LETTER OMEGA"), color = 'b')

```
