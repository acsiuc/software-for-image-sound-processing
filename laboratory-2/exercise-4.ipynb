```python
#EXERCISE 4
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

f0 = 1000 # central frequency (i think)
fs = 48000 #sampling frequency
no_esant = 480 # number of samples

def gen_sinus(f0, fs, no_esant):
    t = np.arange(no_esant)/fs # array with the sample period(Ts) 
    sinusoid = np.sin(2*np.pi*t*f0) # formula for the sinusoid using the Ts
    return sinusoid

sinusoidal_signal = gen_sinus(f0, fs, no_esant)
fft_sinus = np.fft.fft(sinusoidal_signal)

plt.plot(np.abs(fft_sinus))
plt.grid()
plt.show()
```

```python
def gen_rectangular(f0, fs, no_esant):
    t = np.arange(no_esant)/fs
    rectangular = signal.square(2*np.pi*f0*t)

    return rectangular

rectangular_signal = gen_rectangular(f0, fs, no_esant)
fft_rectangular = np.fft.fft(rectangular_signal)
plt.plot(np.abs(fft_rectangular))
plt.grid()
plt.show()
```
