```python
#EXERCISE 5
from pathlib import Path
from PIL import Image
import matplotlib.pyplot as plt

img_path = Path.home() / 'apple.jpg'

def image_conversion(path, saving_name = 'greyscale.png'):
    image = Image.open(path).convert('L')

    plt.imshow(image, cmap = 'gray')
    plt.show()

    image.save(path.parent / saving_name)

    return image

image_conversion(img_path)
```
