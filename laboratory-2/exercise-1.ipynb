```python
#EXERCISE 1

import numpy as np

f0 = 1000 # central frequency (i think)
fs = 48000 #sampling frequency
no_esant = 480 # number of samples

def gen_sinus(f0, fs, no_esant):
    t = np.arange(no_esant)/fs # array with the sample period(Ts) 
    sinusoid = np.sin(2*np.pi*t*f0) # formula for the sinusoid using the Ts
    return sinusoid

samples = gen_sinus(f0, fs, no_esant)

np.savetxt('samples.txt', samples, delimiter=";")
load_samples = np.loadtxt('samples.txt', delimiter=';')

print(load_samples)

```

```python
#EXERCISE 2

from scipy import signal
import matplotlib.pyplot as plt

def gen_rectangular(f0, fs, no_esant):
    t = np.arange(no_esant)/fs
    rectangular = signal.square(2*np.pi*f0*t)
    plt.plot(rectangular)
    plt.ylim(-2, 2)
    plt.grid()
    plt.show()

gen_rectangular(f0, fs, no_esant)
```
