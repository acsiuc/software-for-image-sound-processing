```python
#EXERCISE 2
import numpy as np
from scipy import signal
import matplotlib.pyplot as plt

fs = 48000
ff = 1000
Ts = 1/fs
time_vector = np.arange(480)*Ts
square_wave = signal.square(2*np.pi*ff*time_vector, duty = 0.5)
plt.plot(time_vector, square_wave)
plt.show()

coefficients = np.ones(5)/5

filtered_wave = signal.lfilter(coefficients, 1, square_wave)
```

```python
plt.plot(time_vector, filtered_wave)
plt.show()
```

```python
#EXERCISE 3
filter_coeff = signal.firwin(101, 2000, fs = fs)
```

```python
new_filtered_wave = signal.lfilter(filter_coeff, 1, square_wave)
plt.plot(time_vector, new_filtered_wave)
plt.show()
```
